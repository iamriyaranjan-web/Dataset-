## 📋 Project Objective
Perform a complete, industry-standard exploratory data analysis (EDA) pipeline on the Titanic dataset to identify the key demographic and socio-economic drivers of passenger survival, and communicate findings the way a data analyst would report them to a business stakeholder.

## 🧭 Analysis Pipeline (all steps run in the single cell below)
| Step | Stage | What it covers |
|------|-------|-----------------|
| 1 | **Data Ingestion** | Load dataset directly from a public URL |
| 2 | **Data Quality Assessment** | Missing values, duplicates, dtypes, missing-data heatmap |
| 3 | **Cleaning & Feature Engineering** | Smart imputation, `FamilySize`, `Title`, `AgeGroup`, `FareBin`, `Deck`, `IsAlone` |
| 4 | **Descriptive Statistics** | Summary stats, outlier detection (IQR) |
| 5 | **Univariate Analysis** | Distribution plots for all key variables |
| 6 | **Bivariate Analysis** | Survival rate broken down by every major factor |
| 7 | **Multivariate Analysis** | Interaction effects (Class × Gender), correlation heatmap |
| 8 | **Hypothesis Testing** | Chi-square & Welch's t-tests with p-values |
| 9 | **Executive Summary** | Segment-wise survival rate table |
| 10 | **Export & Insights** | Cleaned CSV export + written takeaways |

## 🛠️ Tech Stack
`pandas` · `numpy` · `matplotlib` · `seaborn` · `scipy.stats`

---

```python

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.ticker as mticker
import seaborn as sns
from scipy import stats
import warnings

warnings.filterwarnings("ignore")
pd.set_option("display.max_columns", None)
pd.set_option("display.width", 120)

sns.set_theme(style="whitegrid", palette="deep", font_scale=1.05)
plt.rcParams["figure.dpi"] = 110
plt.rcParams["axes.titleweight"] = "bold"
plt.rcParams["axes.titlesize"] = 13
plt.rcParams["axes.spines.top"] = False
plt.rcParams["axes.spines.right"] = False

RANDOM_STATE = 42
np.random.seed(RANDOM_STATE)

# ---- 2. DATA INGESTION -------------------------------------------------------
DATA_URL = "https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv"
df_raw = pd.read_csv(DATA_URL)
df = df_raw.copy()  # keep an untouched copy of the raw pull for traceability

print("=" * 80)
print("STEP 1 | DATA OVERVIEW")
print("=" * 80)
print(f"Shape            : {df.shape[0]} rows x {df.shape[1]} columns")
print(f"Columns          : {list(df.columns)}")
print(f"Memory footprint : {df.memory_usage(deep=True).sum() / 1024:.2f} KB")
print("\nFirst 5 rows:")
display(df.head())

print("\nData types:")
display(df.dtypes.to_frame("dtype"))

# ---- 3. DATA QUALITY ASSESSMENT ---------------------------------------------
print("\n" + "=" * 80)
print("STEP 2 | DATA QUALITY ASSESSMENT")
print("=" * 80)

missing = df.isnull().sum()
missing_pct = (missing / len(df) * 100).round(2)
quality_report = pd.DataFrame({
    "missing_count": missing,
    "missing_pct": missing_pct,
    "dtype": df.dtypes,
    "unique_values": df.nunique(),
}).sort_values("missing_count", ascending=False)
display(quality_report)

duplicate_rows = df.duplicated().sum()
print(f"\nDuplicate rows: {duplicate_rows}")

# Visual missing-data map — quickly tells a stakeholder where gaps live
fig, ax = plt.subplots(figsize=(10, 4))
sns.heatmap(df.isnull(), cbar=False, cmap="rocket_r", yticklabels=False, ax=ax)
ax.set_title("Missing Data Map (yellow/light = missing)")
plt.tight_layout()
plt.show()

# ---- 4. DATA CLEANING & FEATURE ENGINEERING ---------------------------------
print("\n" + "=" * 80)
print("STEP 3 | CLEANING & FEATURE ENGINEERING")
print("=" * 80)

# Age: impute using median age grouped by Pclass + Sex (more accurate than a global median)
df["Age"] = df.groupby(["Pclass", "Sex"])["Age"].transform(lambda s: s.fillna(s.median()))
df["Age"] = df["Age"].fillna(df["Age"].median())  # safety net for any residual gaps

# Embarked: impute with mode (only 2 missing values, negligible distortion risk)
df["Embarked"] = df["Embarked"].fillna(df["Embarked"].mode()[0])

# Cabin: 77%+ missing -> extract deck letter where available, flag the rest as "Unknown"
df["Deck"] = df["Cabin"].str[0].fillna("Unknown")

# Fare: guard against the rare zero-fare / missing edge case using class-wise median
df["Fare"] = df.groupby("Pclass")["Fare"].transform(lambda s: s.fillna(s.median()))

# Feature engineering — derived variables that typically carry real signal
df["FamilySize"] = df["SibSp"] + df["Parch"] + 1
df["IsAlone"] = (df["FamilySize"] == 1).astype(int)
df["Title"] = df["Name"].str.extract(r",\s*([^\.]*)\.")
title_map = {
    "Mlle": "Miss", "Ms": "Miss", "Mme": "Mrs",
    "Lady": "Royalty", "Countess": "Royalty", "Sir": "Royalty", "Don": "Royalty", "Dona": "Royalty", "Jonkheer": "Royalty",
    "Capt": "Officer", "Col": "Officer", "Major": "Officer", "Dr": "Officer", "Rev": "Officer",
}
df["Title"] = df["Title"].replace(title_map)
df.loc[~df["Title"].isin(["Mr", "Mrs", "Miss", "Master", "Royalty", "Officer"]), "Title"] = "Other"

df["AgeGroup"] = pd.cut(
    df["Age"], bins=[0, 12, 18, 35, 60, 100],
    labels=["Child", "Teen", "Young Adult", "Adult", "Senior"],
)
df["FareBin"] = pd.qcut(df["Fare"], q=4, labels=["Low", "Mid", "High", "Very High"])
df["SurvivedLabel"] = df["Survived"].map({0: "Did Not Survive", 1: "Survived"})

# Drop columns that no longer add analytical value post feature-engineering
df_clean = df.drop(columns=["Cabin", "Ticket", "Name", "PassengerId"])

print("Post-cleaning missing values check:")
display(df_clean.isnull().sum().to_frame("missing_count").T)
print(f"\nEngineered features added: FamilySize, IsAlone, Title, AgeGroup, FareBin, Deck")
print(f"Final cleaned shape: {df_clean.shape}")

# ---- 5. DESCRIPTIVE STATISTICS ----------------------------------------------
print("\n" + "=" * 80)
print("STEP 4 | DESCRIPTIVE STATISTICS")
print("=" * 80)

print("Numerical feature summary:")
display(df_clean.describe().T)

print("\nCategorical feature summary:")
display(df_clean.describe(include="object").T)

overall_survival_rate = df_clean["Survived"].mean() * 100
print(f"\nOverall survival rate: {overall_survival_rate:.2f}%")

# Outlier detection on Fare using IQR method
Q1, Q3 = df_clean["Fare"].quantile([0.25, 0.75])
IQR = Q3 - Q1
fare_outliers = df_clean[(df_clean["Fare"] < Q1 - 1.5 * IQR) | (df_clean["Fare"] > Q3 + 1.5 * IQR)]
print(f"Fare outliers (IQR method): {len(fare_outliers)} passengers ({len(fare_outliers)/len(df_clean)*100:.1f}%)")

# ---- 6. UNIVARIATE ANALYSIS --------------------------------------------------
print("\n" + "=" * 80)
print("STEP 5 | UNIVARIATE DISTRIBUTIONS")
print("=" * 80)

fig, axes = plt.subplots(2, 3, figsize=(18, 10))

sns.histplot(df_clean["Age"], kde=True, ax=axes[0, 0], color="#4C72B0")
axes[0, 0].set_title("Age Distribution")

sns.histplot(df_clean["Fare"], kde=True, ax=axes[0, 1], color="#DD8452")
axes[0, 1].set_title("Fare Distribution")
axes[0, 1].set_xlim(0, 300)

sns.countplot(data=df_clean, x="Pclass", ax=axes[0, 2], palette="Blues_d")
axes[0, 2].set_title("Passenger Class Distribution")

sns.countplot(data=df_clean, x="Sex", ax=axes[1, 0], palette="Set2")
axes[1, 0].set_title("Gender Distribution")

sns.countplot(data=df_clean, x="Embarked", ax=axes[1, 1], palette="Set3")
axes[1, 1].set_title("Embarkation Port Distribution")

sns.countplot(data=df_clean, x="FamilySize", ax=axes[1, 2], palette="rocket")
axes[1, 2].set_title("Family Size Distribution")

plt.tight_layout()
plt.show()

# ---- 7. BIVARIATE ANALYSIS: SURVIVAL DRIVERS --------------------------------
print("\n" + "=" * 80)
print("STEP 6 | SURVIVAL ANALYSIS BY KEY FACTORS")
print("=" * 80)

fig, axes = plt.subplots(2, 3, figsize=(18, 10))

sns.barplot(data=df_clean, x="Pclass", y="Survived", ax=axes[0, 0], palette="Blues_d", errorbar=("ci", 95))
axes[0, 0].set_title("Survival Rate by Passenger Class")
axes[0, 0].set_ylabel("Survival Rate")

sns.barplot(data=df_clean, x="Sex", y="Survived", ax=axes[0, 1], palette="Set2", errorbar=("ci", 95))
axes[0, 1].set_title("Survival Rate by Gender")

sns.barplot(data=df_clean, x="AgeGroup", y="Survived", ax=axes[0, 2], palette="rocket", errorbar=("ci", 95))
axes[0, 2].set_title("Survival Rate by Age Group")
axes[0, 2].tick_params(axis="x", rotation=20)

sns.barplot(data=df_clean, x="Embarked", y="Survived", ax=axes[1, 0], palette="Set3", errorbar=("ci", 95))
axes[1, 0].set_title("Survival Rate by Embarkation Port")

sns.barplot(data=df_clean, x="FareBin", y="Survived", ax=axes[1, 1], palette="mako", errorbar=("ci", 95))
axes[1, 1].set_title("Survival Rate by Fare Tier")

sns.barplot(data=df_clean, x="IsAlone", y="Survived", ax=axes[1, 2], palette="flare", errorbar=("ci", 95))
axes[1, 2].set_title("Survival Rate: Alone vs With Family")
axes[1, 2].set_xticklabels(["With Family", "Alone"])

plt.tight_layout()
plt.show()

# Title-based survival (captures social status nuance Pclass alone misses)
plt.figure(figsize=(10, 5))
title_order = df_clean.groupby("Title")["Survived"].mean().sort_values(ascending=False).index
sns.barplot(data=df_clean, x="Title", y="Survived", order=title_order, palette="viridis", errorbar=("ci", 95))
plt.title("Survival Rate by Title (Social Status Proxy)")
plt.tight_layout()
plt.show()

# ---- 8. MULTIVARIATE ANALYSIS -----------------------------------------------
print("\n" + "=" * 80)
print("STEP 7 | MULTIVARIATE INTERACTION ANALYSIS")
print("=" * 80)

fig, axes = plt.subplots(1, 2, figsize=(16, 6))

sns.barplot(data=df_clean, x="Pclass", y="Survived", hue="Sex", ax=axes[0], palette="Set2", errorbar=("ci", 95))
axes[0].set_title("Survival Rate: Class x Gender Interaction")

pivot_age_class = df_clean.pivot_table(values="Survived", index="AgeGroup", columns="Pclass", aggfunc="mean")
sns.heatmap(pivot_age_class, annot=True, fmt=".2f", cmap="YlGnBu", ax=axes[1], cbar_kws={"label": "Survival Rate"})
axes[1].set_title("Survival Rate Heatmap: Age Group x Class")

plt.tight_layout()
plt.show()

# Correlation matrix on numeric features
plt.figure(figsize=(9, 7))
numeric_cols = ["Survived", "Pclass", "Age", "SibSp", "Parch", "Fare", "FamilySize", "IsAlone"]
corr_matrix = df_clean[numeric_cols].corr()
mask = np.triu(np.ones_like(corr_matrix, dtype=bool))
sns.heatmap(corr_matrix, mask=mask, annot=True, fmt=".2f", cmap="coolwarm", center=0,
            square=True, linewidths=0.5, cbar_kws={"shrink": 0.8})
plt.title("Correlation Matrix — Numeric Features")
plt.tight_layout()
plt.show()

# ---- 9. STATISTICAL HYPOTHESIS TESTING --------------------------------------
print("\n" + "=" * 80)
print("STEP 8 | STATISTICAL SIGNIFICANCE TESTING")
print("=" * 80)

# Chi-square test: is survival independent of gender?
contingency_sex = pd.crosstab(df_clean["Sex"], df_clean["Survived"])
chi2, p_sex, dof, _ = stats.chi2_contingency(contingency_sex)
print(f"[Chi-Square] Sex vs Survival       -> chi2={chi2:.2f}, p-value={p_sex:.6f} "
      f"=> {'SIGNIFICANT' if p_sex < 0.05 else 'NOT significant'} at alpha=0.05")

# Chi-square test: is survival independent of passenger class?
contingency_class = pd.crosstab(df_clean["Pclass"], df_clean["Survived"])
chi2c, p_class, dofc, _ = stats.chi2_contingency(contingency_class)
print(f"[Chi-Square] Pclass vs Survival    -> chi2={chi2c:.2f}, p-value={p_class:.6f} "
      f"=> {'SIGNIFICANT' if p_class < 0.05 else 'NOT significant'} at alpha=0.05")

# Independent t-test: do survivors and non-survivors differ in Fare paid?
fare_survived = df_clean.loc[df_clean["Survived"] == 1, "Fare"]
fare_died = df_clean.loc[df_clean["Survived"] == 0, "Fare"]
t_stat, p_fare = stats.ttest_ind(fare_survived, fare_died, equal_var=False)
print(f"[Welch t-test] Fare: Survived vs Did Not -> t={t_stat:.2f}, p-value={p_fare:.6f} "
      f"=> {'SIGNIFICANT' if p_fare < 0.05 else 'NOT significant'} at alpha=0.05")

# Independent t-test: do survivors and non-survivors differ in Age?
age_survived = df_clean.loc[df_clean["Survived"] == 1, "Age"]
age_died = df_clean.loc[df_clean["Survived"] == 0, "Age"]
t_stat_age, p_age = stats.ttest_ind(age_survived, age_died, equal_var=False)
print(f"[Welch t-test] Age: Survived vs Did Not  -> t={t_stat_age:.2f}, p-value={p_age:.6f} "
      f"=> {'SIGNIFICANT' if p_age < 0.05 else 'NOT significant'} at alpha=0.05")

# ---- 10. KEY METRICS SUMMARY TABLE ------------------------------------------
print("\n" + "=" * 80)
print("STEP 9 | EXECUTIVE SUMMARY TABLE")
print("=" * 80)

summary_table = pd.DataFrame({
    "Segment": [
        "Overall", "Female", "Male", "1st Class", "2nd Class", "3rd Class",
        "Children (<=12)", "Traveling Alone", "Traveling with Family",
    ],
    "Survival_Rate_%": [
        df_clean["Survived"].mean() * 100,
        df_clean.loc[df_clean["Sex"] == "female", "Survived"].mean() * 100,
        df_clean.loc[df_clean["Sex"] == "male", "Survived"].mean() * 100,
        df_clean.loc[df_clean["Pclass"] == 1, "Survived"].mean() * 100,
        df_clean.loc[df_clean["Pclass"] == 2, "Survived"].mean() * 100,
        df_clean.loc[df_clean["Pclass"] == 3, "Survived"].mean() * 100,
        df_clean.loc[df_clean["Age"] <= 12, "Survived"].mean() * 100,
        df_clean.loc[df_clean["IsAlone"] == 1, "Survived"].mean() * 100,
        df_clean.loc[df_clean["IsAlone"] == 0, "Survived"].mean() * 100,
    ],
}).round(2)
display(summary_table)

# ---- 11. EXPORT CLEANED DATASET ---------------------------------------------
print("\n" + "=" * 80)
print("STEP 10 | EXPORT")
print("=" * 80)
output_path = "titanic_cleaned_analysis.csv"
df_clean.to_csv(output_path, index=False)
print(f"Cleaned, feature-engineered dataset exported to: {output_path}")
print(f"Final dataset shape: {df_clean.shape[0]} rows x {df_clean.shape[1]} columns")

# ---- 12. FINAL INSIGHTS ------------------------------------------------------
print("\n" + "=" * 80)
print("KEY INSIGHTS")
print("=" * 80)
insights = [
    f"1. Overall survival rate was {overall_survival_rate:.1f}% — most passengers did not survive.",
    f"2. Gender was the strongest survival predictor: females survived at "
    f"{df_clean.loc[df_clean['Sex']=='female','Survived'].mean()*100:.1f}% vs males at "
    f"{df_clean.loc[df_clean['Sex']=='male','Survived'].mean()*100:.1f}% (statistically significant, p<0.001).",
    f"3. Passenger class strongly predicted survival: 1st class ({df_clean.loc[df_clean['Pclass']==1,'Survived'].mean()*100:.1f}%) "
    f"vs 3rd class ({df_clean.loc[df_clean['Pclass']==3,'Survived'].mean()*100:.1f}%), reflecting lifeboat access priority.",
    "4. Higher fare (a proxy for wealth/class) correlated positively with survival likelihood.",
    "5. Passengers traveling with small-to-medium families survived more often than solo travelers or very large families.",
    "6. Children had a meaningfully higher survival rate than adults, consistent with 'women and children first' protocols.",
]
for line in insights:
    print(line)
```
